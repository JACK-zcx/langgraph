```python
from langgraph import graph
from langgraph.graph import StateGraph,END,START
from typing import Dict,TypedDict,List
import random
```

```python
class AgentState(TypedDict):
    name: str
    number: list[int]
    counter: int
```

```python
def greeting_node(state: AgentState) -> AgentState:
    '''greet'''
    state["name"] = f'Hi there! {state["name"]}'
    state["counter"] = 0
    return state

def random_node(state: AgentState) -> AgentState:
    '''random'''
    state["number"].append(random.randint(1,10))
    state["counter"] += 1
    return state

def loop(state: AgentState) -> AgentState:
    '''loop'''
    if state["counter"] <= 5:
        print('enter loop',state["counter"])
        return 'loop'
    else:
        return 'exit'
```

```python
graph = StateGraph(AgentState)
graph.add_node('greeting', greeting_node)
graph.add_node('random', random_node)

graph.set_entry_point('greeting')
graph.add_edge('greeting', 'random')
graph.add_conditional_edges('random',loop,{'loop':'random','exit':END})
```

```python
app = graph.compile()
ans = app.invoke({'name':'JACK','number':[],'counter':-2})
print(ans)
```
